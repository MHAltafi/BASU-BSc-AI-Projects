```python
from PIL import Image, ImageDraw, ImageFont
import os
import arabic_reshaper
from bidi.algorithm import get_display

font_paths = []
for (dirpath, dirnames, filenames) in os.walk("./Fonts"):
    for filename in filenames:
        if filename.lower().endswith('.ttf') or filename.lower().endswith('.otf'):
            font_paths.append(os.path.join(dirpath, filename))
    break

texts = [
    "آذربایجان شرقی", "آذربایجان غربی", "اردبیل", "اصفهان", "البرز", "ایلام",
    "بوشهر", "تهران", "چهارمحال و بختیاری", "خراسان جنوبی", "خراسان رضوی", "خراسان شمالی",
    "خوزستان", "زنجان", "سمنان", "سیستان و بلوچستان", "فارس", "قزوین", "قم",
    "کردستان", "کرمان", "کرمانشاه", "کهگیلویه و بویراحمد", "گلستان", "گیلان",
    "لرستان", "مازندران", "مرکزی", "هرمزگان", "همدان", "یزد"
]

if not os.path.exists("output"):
    os.makedirs("output")

counter = 0
for font_path in font_paths:
    counter += 1
    print(f'{counter} : {font_path}')
    for text in texts:
        try:
            font = ImageFont.truetype(font_path, size=60)
            
            reshaped_text = arabic_reshaper.reshape(text)
            bidi_text = get_display(reshaped_text)
            
            temp_image = Image.new("L", (1, 1), 255)
            draw = ImageDraw.Draw(temp_image)
            
            bbox = draw.textbbox((0, 0), bidi_text, font=font)
            text_width = bbox[2] - bbox[0]
            text_height = bbox[3] - bbox[1]
            
            margin = 30
            image_width = text_width + 4 * margin
            image_height = text_height + 4 * margin
            
            image = Image.new("L", (image_width, image_height), 255)
            draw = ImageDraw.Draw(image)
            
            text_x = margin
            text_y = margin
            draw.text((text_x, text_y), bidi_text, fill=0, font=font)
            
            output_path = f"output/{counter}_{text}.png"
            image.save(output_path)
        except Exception as e:
            print(f"Error in loading font {font_path}: {e}")

print("success!")
```

```python
def find_bounding_box(img):
    pixels = img.load()
    width, height = img.size
    
    left = width
    right = 0
    top = height
    bottom = 0
    
    for y in range(height):
        for x in range(width):
            if pixels[x, y] == 0:
                if x < left:
                    left = x
                if x > right:
                    right = x
                if y < top:
                    top = y
                if y > bottom:
                    bottom = y
    
    if left == width or right == 0 or top == height or bottom == 0:
        return None
    
    return (left, top, right, bottom)

def crop_image_with_padding(image_path, output_path, padding=10):
    img = Image.open(image_path).convert("L")
    
    bbox = find_bounding_box(img)
    
    if bbox:
        left = max(0, bbox[0] - padding)
        top = max(0, bbox[1] - padding)
        right = min(img.width, bbox[2] + padding)
        bottom = min(img.height, bbox[3] + padding)
        
        cropped_img = img.crop((left, top, right, bottom))
        
        cropped_img.save(output_path)
    else:
        print(f"No content found in image: {image_path}")

def process_images(input_folder='output', output_folder='output2', padding=10):
    os.makedirs(output_folder, exist_ok=True)
    
    for filename in os.listdir(input_folder):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.gif')):
            input_image_path = os.path.join(input_folder, filename)
            output_image_path = os.path.join(output_folder, filename)
            
            crop_image_with_padding(input_image_path, output_image_path, padding)
            print(f"Processed and saved: {output_image_path}")
        else:
            print(f"Skipping non-image file: {filename}")

process_images()
```
